# Reranker — Analysis Notebook

This notebook evaluates the reranker model using:
- Sample candidate lists
- Ranking quality metrics
- Mean Reciprocal Rank (MRR)
- Normalized Discounted Cumulative Gain (nDCG)

```python
import numpy as np

def mock_reranker(candidates):
    # Sort by length as a placeholder
    return sorted(candidates, key=lambda x: len(x), reverse=True)

sample_candidates = [
    ['hello world', 'hi', 'greetings'],
    ['help me', 'assist', 'support request'],
    ['what is ai', 'define ai', 'explain artificial intelligence']
]
```

```python
def mrr(ranked):
    return 1.0 / (ranked.index(ranked[0]) + 1)

def ndcg(ranked):
    gains = [1.0 / (i + 1) for i in range(len(ranked))]
    return sum(gains) / sum(gains)

results = []
for c in sample_candidates:
    ranked = mock_reranker(c)
    results.append({
        'input': c,
        'ranked': ranked,
        'mrr': mrr(ranked),
        'ndcg': ndcg(ranked)
    })

results
```
